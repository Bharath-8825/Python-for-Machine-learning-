```python
DEF: a loop is a programming concept used to execute
=>A block of code repeatedly until certain condition is met or the sequence is completed

Types of loops
-> for loop
     1) for loop with sequence
     2) for loop with range
-> while loop
-> nested loop
     1)for loop inside for loop
     2)for loop inside while loop
     3)while loop inside while loop
     4)while loop inside for loop
```

```python
=>for loop with sequence syntax

for iterative variable in sequence variable:
    statements
```

```python
name="Bharath"
for ch in name:
    print(ch)
```

```python
name="Bharath"
pos=0
for i in name:
    print(f"position:{pos} character:{i}")
    pos+=1
```

```python
#Method-2 : Enumerate
name="Bharath"
for i in enumerate(name):
    print(i)
    print(f"position:{i[0]} and character:{i[1]}")
```

```python
l1=["priya","karan","kiran","praveen"]
for i in l1:
    print(f"hello good afternoon {i} happy diwali")
```

```python
# For loop with range
=> initialization - start value
   condition - stop value
   inc/dec - step value
=> for loop with range syntax:
   range(start value,stop value,step value)
   default start value is zero(0)
   step default value is 1
=> for iterative variable in range(start,stop,step):
       statements
```

```python
for i in range(1,11,1):
    print(i)
```

```python
for j in range(51,61):
    print(j)
```

```python
for i in range(101,111):
    if i%2==0:
        print("even:",i)
    else:
        print("odd:",i)
```

```python
for i in range(102,111,2):
    print(i)
for k in range(101,111,2):
    print(k)
```
