```python
""" 11. Accept a year. First check whether it is divisible by 4. 
If yes, check whether it is divisible by 100. 
If it is, check whether it is divisible by 400. 
Display whether the year is a leap year."""
year = int(input("Enter year: "))
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap Year")
        else:
            print("Not a Leap Year")
    else:
        print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
""" 12. Accept three side lengths. 
First check whether they can form a valid triangle. 
If valid, check whether all three sides are equal, 
exactly two sides are equal, or all sides are different.
"""
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    if a == b and b == c:
        print("Equilateral Triangle")
    elif a == b or b == c or a == c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
else:
    print("Invalid Triangle")
```

```python
"""13.Accept years of experience and performance rating. 
If experience is at least 5 years, check the rating. 
Give a 20% bonus for an excellent rating and 10% for a good rating. 
If experience is below 5 years, give a 5% bonus only for an excellent rating."""

experience = int(input("Enter years of experience: "))
rating = input("Enter performance rating: ").lower()

if experience >= 5:
    if rating == "excellent":
        print("Bonus: 20%")
    elif rating == "good":
        print("Bonus: 10%")
    else:
        print("Bonus: 0%")
else:
    if rating == "excellent":
        print("Bonus: 5%")
    else:
        print("Bonus: 0%")
```

```python
""" 14. Accept a password. If its length is at least 8 characters,
check whether it contains a digit. If it does, check whether it
contains an uppercase letter. Classify it as strong or moderate based 
on these conditions. Otherwise, classify it as weak.
"""
password = input("Enter password: ")

if len(password) >= 8:
    if any(ch.isdigit() for ch in password):
        if any(ch.isupper() for ch in password):
            print("Strong Password")
        else:
            print("Moderate Password")
    else:
        print("Moderate Password")
else:
    print("Weak Password")
```

```python
"""20. Accept account type, account balance, and transaction type (deposit or withdrawal). 
If the account type is savings, check the transaction type. 
For withdrawals, check minimum balance requirements; for deposits, validate the deposit amount. 
If the account type is current, apply a different minimum balance rule."""
account = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter account balance: ₹"))
transaction = input("Enter transaction (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: ₹"))

if amount <= 0:
    print("Invalid Transaction Amount")
elif account == "savings":
    if transaction == "deposit":
        balance = balance + amount
        print("Deposit Successful")
    elif transaction == "withdrawal":
        if balance - amount >= 1000:
            balance = balance - amount
            print("Withdrawal Successful")
        else:
            print("Minimum Balance Must Be ₹1000")
    else:
        print("Invalid Transaction Type")
elif account == "current":
    if transaction == "deposit":
        balance = balance + amount
        print("Deposit Successful")
    elif transaction == "withdrawal":
        if balance - amount >= 5000:
            balance = balance - amount
            print("Withdrawal Successful")
        else:
            print("Minimum Balance Must Be ₹5000")
    else:
        print("Invalid Transaction Type")
else:
    print("Invalid Account Type")

print("Final Balance: ₹", balance)
```
