```python
#fibonacci series
n1=0
n2=1
i=1
while i<=10:
    print(n1,end=" ")
    s=n1+n2
    n1,n2=n2,s
    i+=1
```

```python
#sum oof digits
# 5+3+8=16
n=int(input()) #initialization
s=0
while(n>0):
    rem = n%10 #8
    s = s+rem
    n = n//10
print(f"the sum of digits in the num is:",s)  
```
