```python
for i in range(1,15,1):
    if(i==8):
        break
    else:
        print(i)
```

```python
for i in range(1,15,1):
    if(i==8):
        continue
    else:
        print(i)
```

```python
for i in range(1,10,1):
    if(i==8):
        pass
    else:
        print(i)
```
