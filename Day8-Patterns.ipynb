```python
while loop inside while loop syntax:

initialization of outer while
while(outer condition):
    initialization of inner while
    while(inner condition):
        statements of inner while
        incre/decre of inner while
    statements of outer while
    incre/decre of outer while
```

```python
i=1
while(i<=5):
    j=1
    while(j<=5):
        print("*",end=" ")
        j+=1
    print()
    i+=1
```

```python
i = 1
while i <= 5:
    j = 1
    while j <= 5:
        if (i == 1 or j == 1 or i == 5 or j == 5):
            print("*", end=" ")
        else:
            print(" ", end=" ")
        j += 1  # Aligned with inner while loop
    print()
    i += 1      # Aligned with outer while loop
```
